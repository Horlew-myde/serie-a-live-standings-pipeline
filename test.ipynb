#### Serie A league Standings using API data 


```python
import os
import json
import requests
import pandas as pd
from mysql import connector
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
API_KEY = os.getenv("API_KEY")
# print(API_KEY)
API_HOST = os.getenv("API_HOST")
category_type = "soccer",
country = "italy",
league = "serie-a"
```

```python
url = "https://livescore6.p.rapidapi.com/leagues/v2/get-table"

querystring = {
    "Category": category_type,
    "Ccd": country,
    "Scd": league
}

headers = {
    "x-rapidapi-key": API_KEY,
    "x-rapidapi-host": API_HOST
}
```

```python
response = requests.get(url=url,   
                         headers=headers,  
                         params= querystring)
```

```python
payload = response.json()
```

```python
payload
```

## ------------------------------ TRANSFORM   ----------------------

```python
formatted_response = json.dumps(payload, indent=4)
```

```python
print(formatted_response)
```

```python
payload["LeagueTable"]
```

```python
#first team on the list
payload["LeagueTable"]["L"][0]["Tables"][0]["team"][0]
```

```python
standing_list = payload["LeagueTable"]["L"][0]["Tables"][0]["team"]
```

```python
formatted_standing_list = json.dumps(standing_list, indent=4)
print(f"\n\n\n\n\{formatted_standing_list}")
```

```python
points = standing_list[1]['pts']
print(points)
```

```python
rows = []
column_names = ['season', 'position', 'team', 'played', 'won', 'draw', 'lost', 'goals_for', 'goals_against', 'goals_difference', 'points' ]

for club in standing_list:
    season = 2025
    position = club["rnk"]
    team = club["Tnm"]
    played = club["pld"]
    won = club["win"]
    draw = club["drw"]
    lost = club["lst"]
    goals_for = club["gf"]
    goals_against = club["ga"]
    goals_difference = club["gd"]
    points = club["pts"]

    tuple_of_club_records = (season, position, team, played, won, draw, lost, goals_for, goals_against, goals_difference, points)
    rows.append(tuple_of_club_records)

```

```python
print(rows)
```

```python
# we change it to a dataframe

df = pd.DataFrame(rows, columns=column_names)
```

## ----------------------------- MYSQL PART STARTS HERE ---------------------------

```python
MYSQL_HOST = os.getenv("MYSQL_HOST")
# print(MYSQL_HOST)
MYSQL_PORT = os.getenv("MYSQL_PORT")
MYSQL_USER = os.getenv("MYSQL_USER")
MYSQL_PASSWORD = os.getenv("MYSQL_PASSWORD")
MYSQL_DATABASE = os.getenv("MYSQL_DATABASE")
```

```python
print(MYSQL_USER)
```

```python
server_con = connector.connect(
    host = MYSQL_HOST,
    port = MYSQL_PORT,
    user = MYSQL_USER,
    password = MYSQL_PASSWORD,
    connection_timeout = 10,
    autocommit = False,
    raise_on_warnings = True
)

server_cur = server_con.cursor()
print(f"[SUCCES] - connected to MYSQL server!!!") 
```

```python
server_cur.close()
server_con.close()
```

```python
db_connection = connector.connect(
    host = MYSQL_HOST,
    port = MYSQL_PORT,
    user = MYSQL_USER,
    password = MYSQL_PASSWORD,
    database = MYSQL_DATABASE

)
cur = db_connection.cursor()
print(f"[SUCCESS] - connection to database is successfull!!!")
```

```python
serie_a_standings = "standings"
cur.execute("SHOW TABLES LIKE %s", (f"{serie_a_standings}",))

if cur.fetchone() is None:
    raise SystemExit(f"this table '{serie_a_standings}' is NOT FOUND!!!.... please create it")
else:
    print(f"[SUCCESS] - This table '{serie_a_standings}' exists! continue to the next phase !!!")
```

### ------------------------ now we can start the 'UPSERT' Operation --------------------

```python
table_columns = ['season', 'position', 'team', 'played', 'won', 'draw', 'lost', 'goals_for', 'goals_against', 'goals_difference', 'points' ]

standings_df = df[table_columns]
```

```python
standings_tuples = standings_df.itertuples(index=False, name=None)

list_of_standings_tuples = list(standings_tuples)
```

```python
UPSERT_SQL = f"""
INSERT INTO {serie_a_standings}
(season, position, team, played, won, draw, lost, goals_for, goals_against, goals_difference, points)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s) AS new_data
ON DUPLICATE KEY UPDATE
    position = new_data.position,
    team = new_data.team,
    played = new_data.played,
    won = new_data.won,
    draw = new_data.draw,
    lost = new_data.lost,
    goals_for = new_data.goals_for,
    goals_against = new_data.goals_against,
    goals_difference = new_data.goals_difference,
    points = new_data.points
"""
```

```python
num_of_upsert_rows = len(list_of_standings_tuples)
print(num_of_upsert_rows)
```

```python
try:
    cur.executemany(UPSERT_SQL, list_of_standings_tuples)
    db_connection.commit()
    print(f"[SUCCESS] - Upsert attempted for {num_of_upsert_rows} rows!!")
except Exception as e:
    db_connection.rollback()
    print(f"[ERROR] - Rolled back due to : {e}")
finally:
    cur.close()
    db_connection.close()
    print("All Database connections are now closed. \n\nclean up completed")
```
